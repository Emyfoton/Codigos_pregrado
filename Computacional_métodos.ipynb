```python
import numpy as np
from scipy.special import sph_harm
from scipy.special import yn
import matplotlib.pyplot as plt
```

# 1. Expansión multipolar y anisotropías del CMB

```python
#combinaciones simples de armónicos esféricos.

phi = np.linspace(0, 2*np.pi, 200)
theta = np.linspace(0, np.pi, 200)

Phi, Theta = np.meshgrid(phi, theta)

m = 1

valores_l = [2, 3, 4, 5]

fig, axs = plt.subplots(2,2)

for ax, l in zip(axs.flat, valores_l):

    Y = sph_harm(m, l, Phi, Theta)

    ax.imshow(np.real(Y), cmap='RdYlBu_r')

    ax.set_title(f"l = {l}, m = {m}")

plt.tight_layout()
plt.show()
```

# 2. Funciones de Neumann

```python
# Gráfico N_n(x)
x=np.linspace(0.01, 20, 1000)
y0= yn(0,x)
y1= yn(1, x)
y2= yn(2,x)


plt.ylim(-0.6,0.6)
plt.axhline(0, c="k")
plt.plot(x, y0, c="brown", label="N_0(x)")
plt.plot(x, y1, c="lightcoral", label="N_1(x)")
plt.plot(x, y2, c="darkkhaki", label="N_2(x)")
plt.grid(c="lightgray")
plt.legend()

plt.show()
```
