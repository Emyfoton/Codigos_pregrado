
a. Grafique combinaciones simples de armónicos esféricos.


```python
import numpy as np
from scipy.special import sph_harm_y
import matplotlib.pyplot as plt

phi = np.linspace(0, 2*np.pi, 200)
theta = np.linspace(0, np.pi, 200)

Theta, Phi = np.meshgrid(phi, theta)

m = 4

valores_l = [1, 2, 3, 4]

fig, axs = plt.subplots(2,2)

for ax, l in zip(axs.flat, valores_l):

    Y = sph_harm_y(m, l, Phi, Theta)

    ax.imshow(np.real(Y), cmap='RdYlBu_r', interpolation='bilinear')

    ax.set_title(f"l = {l}, m = {m}")

plt.tight_layout()
plt.show()
```
