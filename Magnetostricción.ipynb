```python
import numpy as np
import matplotlib.pyplot as plt
```

# Datos Hierro

```python
#Polarización Positiva
#Min a Máx

I1_Fe= np.array([0.15, 0.66, 1.38, 3.44])
N1_Fe= np.array([0, 1, 2, 3])

plt.scatter(I1_Fe, N1_Fe)
plt.title("Corriente en polaridad positiva Fe de Min a Máx")
plt.xlabel("Número de mínimos")
plt.ylabel("I (A)")
```
